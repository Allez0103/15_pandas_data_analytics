```python
import pandas as pd

score_list = [85, 92, 78, 90, 88]
name_index =  ['Alice', 'Bob', 'Charlie', 'David', 'Eve']

Scores = pd.Series(score_list)
print(Scores)
print()

Scores = pd.Series(score_list, index=name_index)

dict_scores = {'Python': 95, 'SQL': 85, 'Web': 90}

Scores_2 = pd.Series(dict_scores)

print(Scores)
print()
print(Scores_2)
```
