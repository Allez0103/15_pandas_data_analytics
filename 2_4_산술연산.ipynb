```python
# 시리즈와 시리즈 연산

import pandas as pd

dict_data1 = {
    "국어" : 100,
    "영어" : 80,
    "수학" : 90
}

dict_data2 = {
    "수학" : 80,
    "국어" : 90,
    "영어" : 80
}

student1 = pd.Series(dict_data1)
student2 = pd.Series(dict_data2)

print(student1)
print()
print(student2)
```

```python
addition = student1 + student2
print(type(addition))
print()
print(addition)
```

```python
substraction = student1 - student2
print(type(substraction))
print()
print(substraction)
```

```python
multiplication = student1 * student2
print(type(multiplication))
print()
print(multiplication)
```

```python
division = student1 / student2
print(type(division))
print()
print(division)
```

```python
lis = [addition, substraction, multiplication, division]
df_index = ["덧셈", "뺄셈", "곱셈", "나눗셈"]

result = pd.DataFrame(lis, index=df_index)
result
```

```python
# NaN 값이 있는 시리즈 연산

import pandas as pd
import numpy as np

dict_data1 = {
    "국어" : np.nan,
    "영어" : 80,
    "수학" : 90
}

dict_data2 = {
    "수학" : 80,
    "국어" : 90
}

student1 = pd.Series(dict_data1)
student2 = pd.Series(dict_data2)

print(student1)
print()
print(student2)
```

```python
addition = student1 + student2
print(addition)
```

```python
substraction = student1 - student2
print(substraction)
```

```python
multiplication = student1 * student2
print(multiplication)
```

```python
division = student1 / student2
print(division)
```

```python
result = pd.DataFrame([addition, substraction, multiplication, division], index=df_index)
result
```

```python
# 연산메서드

import pandas as pd
import numpy as np

dict_data1 = {
    "국어" : np.nan,
    "영어" : 80,
    "수학" : 90
}

dict_data2 = {
    "수학" : 80,
    "국어" : 90
}

student1 = pd.Series(dict_data1)
student2 = pd.Series(dict_data2)

sr_add = student1.add(student2, fill_value=0)
sr_sub = student1.sub(student2, fill_value=0)
sr_mul = student1.mul(student2, fill_value=0)
sr_div = student1.div(student2, fill_value=0)

lis = [sr_add, sr_sub, sr_mul, sr_div]

result = pd.DataFrame(lis, index=df_index)
result
```

```python
# 데이터프레임과 데이터프레임 연산

import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")
df = titanic.loc[:, ["age", "fare"]]

df.tail(10)
```

```python
addition = df + 10
addition.tail(10)
```

```python
substraction = addition - df
substraction.tail(10)
type(substraction)
```

```python
sample1 = addition.tail(10)
print(sample1.shape)
print()
print(sample1)
```

```python
sample2 = substraction.tail(10).fillna(0)
print(sample2.shape)
print()
print(sample2)
```

```python
df_add = sample1.add(sample2, fill_value=0)
df_sub = sample1.sub(sample2, fill_value=0)
df_mul = sample1.mul(sample2, fill_value=0)
df_div = sample1.div(sample2, fill_value=0)

print(df_add)
print()
print(df_sub)
print()
print(df_mul)
print()
print(df_div)
```
