```python
# 회귀선이 있는 산점도

import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset('titanic')

sns.set_style('darkgrid')

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.regplot(
    x='age',
    y='fare',
    data=titanic,
    ax=axes[0]
)

sns.regplot(
    x='age',
    y='fare',
    data=titanic,
    ax=axes[1],
    fit_reg=False
)

plt.show()
```

```python
# 히스토그램 / 커널 밀도 그래프

fig, axes = plt.subplot_mosaic(
    [
        ['top_left', 'top_right'],
        ['middle_left', 'middle_right'],
        ['bottom', 'bottom']
    ],
    figsize=(15, 6),
    constrained_layout=False
)

sns.histplot(x='age', data=titanic, ax=axes['top_left'])
sns.histplot(x='age',hue='survived', data=titanic, ax=axes['top_right'])
sns.histplot(x='age',hue='survived', multiple='dodge', data=titanic, ax=axes['middle_left'])
sns.histplot(x='age',hue='survived', multiple='stack', data=titanic, ax=axes['middle_right'])
sns.histplot(x='age',hue='survived', multiple='fill', bins=10, data=titanic, ax=axes['bottom'])

axes['top_left'].set_title('Histogram')
axes['top_right'].set_title('Histogram (hue)')
axes['middle_left'].set_title('Histogram (multiple - dodge)')
axes['middle_right'].set_title('Histogram (multiple - stack)')
axes['bottom'].set_title('Histogram (multiple - fill)')

plt.show()
```

```python
# 커널 밀도 그래프

fig, axes = plt.subplot_mosaic(
    [
        ['top_left', 'top_center', 'right'],
        ['bottom_left', 'bottom_center', 'right'],
    ],
    figsize=(15, 6),
    constrained_layout=True
)

sns.kdeplot(x='age', data=titanic, ax=axes['top_left'])
sns.kdeplot(x='age', data=titanic, ax=axes['bottom_left'], hue='survived')
sns.kdeplot(x='age', data=titanic, ax=axes['top_center'], hue='survived', fill=True)
sns.kdeplot(x='age', data=titanic, ax=axes['bottom_center'], hue='survived', multiple='stack')
sns.kdeplot(x='age', data=titanic, ax=axes['right'], hue='survived', multiple='fill', bw_adjust=2.0)

fig.suptitle("Titanic - Age Distribution", size=20)

axes['top_left'].set_title('KDE')
axes['bottom_left'].set_title('KDE (hue)')
axes['top_center'].set_title('KDE (fill=True)')
axes['bottom_center'].set_title('KDE (multiple - stack)')
axes['right'].set_title('KDE (multiple - fill)')

plt.show()
```

```python
# 히트맵

import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset('titanic')

sns.set_style('darkgrid')

table = titanic.pivot_table(index=['sex'], columns=['class'], aggfunc='size', observed=False)
# print(table)

sns.heatmap(
    table,
    annot=True,
    fmt='d',
    cmap='YlGnBu',
    linewidths=.5,
    cbar=True
)

plt.show()
```

```python
# 범주형 데이터의 산점도

import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset('titanic')

sns.set_style('whitegrid')

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.stripplot(
    data=titanic,
    x='class',
    y='age',
    ax=axes[0]
)

sns.swarmplot(
    data=titanic,
    x='class',
    y='age',
    ax=axes[1],
    hue='class',
    size=4
)

axes[0].set_title('Strip Plot')
axes[1].set_title('Swarm Plot')

plt.show()
```

```python
# 막대 그래프

sns.set_style('ticks')

fig, axes = plt.subplots(2, 2, figsize=(10, 6), constrained_layout=True)

sns.barplot(
    data=titanic,
    x='sex',
    y='survived',
    ax=axes[0, 0]
)

sns.barplot(
    data=titanic,
    x='sex',
    y='survived',
    ax=axes[0, 1],
    errorbar=None
)

sns.barplot(
    data=titanic,
    x='sex',
    y='survived',
    ax=axes[1, 0],
    hue='class',
    errorbar=('ci', 95),
    estimator='median'
)

sns.barplot(
    data=titanic,
    x='sex',
    y='survived',
    ax=axes[1, 1],
    hue='class',
    dodge=False
)

axes[0, 0].set_title('titanic survived - sex')
axes[0, 1].set_title('titanic survived - sex(w/o errorbar)')
axes[1, 0].set_title('titanic survived - sex/class(median)')
axes[1, 1].set_title('titanic survived - sex/class(dodge)')

plt.show()
```

```python
# 빈도 그래프

import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset('titanic')

sns.set_style('whitegrid')

fig, axes = plt.subplots(1, 4, figsize=(15, 5))

sns.countplot(
    data=titanic,
    x='class',
    ax=axes[0]
)

sns.countplot(
    data=titanic,
    x='class',
    ax=axes[1],
    hue='who',
    palette='Set1',
    dodge=True
)

sns.countplot(
    data=titanic,
    x='class',
    ax=axes[2],
    hue='who',
    palette='Set2',
    dodge=False
)

sns.countplot(
    data=titanic,
    x='class',
    ax=axes[3],
    hue='class',
    palette='Set3'
)

axes[0].set_title('titanic class')
axes[1].set_title('titanic class - who')
axes[2].set_title('titanic class - who(dodge)')
axes[3].set_title('titanic class - color')

plt.show()
```

```python
# 박스 플롯 / 바이올린 그래프

import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset('titanic')

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

sns.boxplot(
    data=titanic,
    x='alive',
    y='age',
    ax=axes[0, 0]
)

sns.boxplot(
    data=titanic,
    x='alive',
    y='age',
    ax=axes[0, 1],
    hue='sex'
)

sns.violinplot(
    data=titanic,
    x='alive',
    y='age',
    ax=axes[1, 0]
)

sns.violinplot(
    data=titanic,
    x='alive',
    y='age',
    ax=axes[1, 1],
    hue='sex'
)

plt.show()
```

```python
# 조인트 그래프

import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset('titanic')

j1 = sns.jointplot(data=titanic, x='fare', y='age')
j2 = sns.jointplot(data=titanic, x='fare', y='age', kind='reg')
j3 = sns.jointplot(data=titanic, x='fare', y='age', kind='hex')
j4 = sns.jointplot(data=titanic, x='fare', y='age', kind='kde')

j1.figure.suptitle('titanic fare - scatter', size=15)
j2.figure.suptitle('titanic fare - reg', size=15)
j3.figure.suptitle('titanic fare - hex', size=15)
j4.figure.suptitle('titanic fare - kde', size=15)

plt.show()
```
