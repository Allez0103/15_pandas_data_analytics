```python
# csv 파일로 저장

import pandas as pd

data = {
    "name" : ["Jerry", "Riah", "Paul"],
    "algol" : ["A", "A+", "B"],
    "basic" : ["C", "B", "B+"],
    "c++" : ["B+", "C", "C+"]
}

df = pd.DataFrame(data)
df = df.set_index("name")
df
```

```python
df.to_csv("./data/df_sample2.csv")
```

```python
df.to_csv("./data/df_sample_no_index2.csv", index=False)
```

```python
# json 파일로 저장

import pandas as pd

data = {
    "name" : ["Jerry", "Riah", "Paul"],
    "algol" : ["A", "A+", "B"],
    "basic" : ["C", "B", "B+"],
    "c++" : ["B+", "C", "C+"]
}

df = pd.DataFrame(data)
df = df.set_index("name")
df
```

```python
df.to_json("./data/df_sample2.json")
```

```python
df.to_json("./data/df_sample_split2.json", orient="split")
```

```python
df.to_json("./data/df_sample_index2.json", orient="index")
```

```python
df.to_json("./data/to_sample_columns2.json", orient="columns")
```

```python
df.to_json("./data/df_sample_values2.json", orient="values")
```
