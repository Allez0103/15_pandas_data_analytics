```python
# 열 추가
# DF객체["추가하려는 열 이름"] = 데이터 값

exam_data = {
    "이름" : ["서준", "우현", "인아"],
    "수학" : [90, 80, 70],
    "영어" : [98, 89, 95],
    "음악" : [95, 95, 100],
    "체육" : [100, 90, 90]
}

df = pd.DataFrame(exam_data)
df["국어"] = 80
df["미술"] = [80, 90, 100]
df
```

```python
# 행 추가
# DF객체.loc["추가할 행 이름"] = 데이터 값 또는 배열

exam_data = {
    "이름" : ["서준", "우현", "인아"],
    "수학" : [90, 80, 70],
    "영어" : [98, 89, 95],
    "음악" : [95, 95, 100],
    "체육" : [100, 90, 90]
}

df = pd.DataFrame(exam_data)
df.loc[3] = 0
df.loc[4] = ["동규", 90, 80, 70, 60]
df.loc["행5"] = df.loc[1]
df.loc["행6"] = df.loc[3]
df
```

```python
# 원소 값 변경

exam_data = {
    "이름" : ["서준", "우현", "인아"],
    "수학" : [90, 80, 70],
    "영어" : [98, 89, 95],
    "음악" : [95, 95, 100],
    "체육" : [100, 90, 90]
}

df = pd.DataFrame(exam_data)
df = df.set_index("이름")
print(df)
print()
df.iloc[0, 3] = 95
df.loc["서준", "체육"] = 90
print(df)
```

```python
print(df.iloc[0].iloc[3])
print(df.iloc[0]["체육"])
print(df.loc["서준"]["체육"])
```

```python
# 행과 열의 위치 바꾸기
# DF객체.transpose()
# DF객체.T

exam_data = {
    "이름" : ["서준", "우현", "인아"],
    "수학" : [90, 80, 70],
    "영어" : [98, 89, 95],
    "음악" : [95, 95, 100],
    "체육" : [100, 90, 90]
}

df = pd.DataFrame(exam_data)

df = df.transpose()
print(df)
print()

df = df.T
print(df)
```
