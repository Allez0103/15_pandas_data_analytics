```python
# 행인덱스 초기화

import pandas as pd

dict_data = {
    "c0" : [1, 2, 3],
    "c1" : [4, 5, 6],
    "c2" : [7, 8, 9],
    "c3" : [10, 11, 12],
    "c4" : [13, 14, 15]
}

df = pd.DataFrame(dict_data, index=["r0", "r1", "r2"])
print(df)
```

```python
ndf = df.reset_index()
print(df)
print()
print(ndf)
```

```python
ndf2 = df.reset_index(names=["c00"])
print(ndf2)
```

```python
ndf3 = df.reset_index(drop=True)
print(ndf3)
```

```python
# 행인덱스를 기준으로 데이터 프레임 정렬
# sort_index()
# DF객체.sort_index()

import pandas as pd

dict_data = {
    "c0" : [1, 2, 3],
    "c1" : [4, 5, 6],
    "c2" : [7, 8, 9],
    "c3" : [10, 11, 12],
    "c4" : [13, 14, 15]
}

df = pd.DataFrame(dict_data, index=["r0", "r2", "r1"])
print(df)
print()

ndf = df.sort_index()
print(ndf)
```

```python
# 인덱스 오름차순/내림차순 정렬

ndf2 = df.sort_index(ascending=False)
print(ndf2)
print()

ndf3 = df.sort_index(ascending=True)
print(ndf3)
```

```python
# 열기준 정렬
# DF객체.sort_values()

import pandas as pd

dict_data = {
    "c0" : [1, 2, 3],
    "c1" : [4, 5, 6],
    "c2" : [7, 8, 9],
    "c3" : [10, 11, 12],
    "c4" : [13, 14, 15]
}

df = pd.DataFrame(dict_data, index=["r0", "r1", "r2"])
print(df)
print()

ndf = df.sort_values(by="c2", ascending=False)
print(ndf)
```

```python
ndf2 = df.sort_values(by=["c3", "c4"], ascending=[False, True])
print(ndf2)
```

```python
import pandas as pd

dict_data = {
    "c0" : [1, 2, 3],
    "c1" : [4, 5, 6],
    "c2" : [7, 8, 9],
    "c3" : [10, 11, 11],
    "c4" : [13, 14, 15]
}

df = pd.DataFrame(dict_data, index=["r0", "r1", "r2"])

ndf2 = df.sort_values(by=["c3", "c4"], ascending=[False, True])
print(ndf2)

# c3에 중복값이 있을 경우 c4를 기준으로 정렬
```
