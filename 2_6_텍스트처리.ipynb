```python
# 문자열 메서드

import pandas as pd
import numpy as np

list_data = ["Apple_사과", "Banana_바나나", "Cherry_체리", np.nan]
ser = pd.Series(list_data, index=["First ", "Second ", "Third ", "Fourth"], dtype="string")

print(ser)
```

```python
# 소문자로 변환 -> str접근자.lower()

print(ser.str.lower())
print()

# 대문자로 변환 -> str접근자.upper()

print(ser.str.upper())
print()

# 문자열 길이 계산 -> str접근자.len()

print(len(ser)) # -> 객체의 행의 개수 계산
print()
print(ser.str.len())
```

```python
# 문자열 분할

exam1 = ser.str.split("_")
print(type(exam1))
print()
print(exam1)
```

```python
# 문자열을 분할하여 데이터프레임으로 확장

exam2 = ser.str.split("_", expand=True)
print(type(exam2))
print()
print(exam2)
```

```python
# 분열된 문자열 중 특정 요소 선택

exam3 = ser.str.split("_").str.get(1)
print(type(exam3))
print()
print(exam3)
print()
print()

exam4 = ser.str.split("_").str.get(0)
print(type(exam4))
print()
print(exam4)
```

```python
# 행인덱스 추출

idx = ser.index
print(idx)
```

```python
print(idx.str.strip())
```

```python
# 문자열 교체

print(ser)
print()
print(ser.str.replace("_", ":", regex=False))
```

```python
print(ser.str.replace("[^a-zA-Z\\s]", "", regex=True))
```

```python
# 접근자 인덱싱 -> str[a]

print(ser.str[0])
```

```python
# 접근자 슬라이싱 -> str[a:b]

print(ser)
print()
print(ser.str[0:4])
```

```python
# 특정 문자 포함 여부 확인

print(ser)
print()

contains_A = ser.str.contains("A", na=False)
print(contains_A)
print()

contains_pattern = ser.str.contains(r"[A|B][a-z]+")
print(contains_pattern)
```

```python
print(ser)
ser.str.fullmatch(r"[A|B][a-z]+")
```

```python
list_data = ["a1", "b2", "c3", "d4a5", "e6e7", "f8"]
ser3 = pd.Series(list_data, dtype="string")
print(ser3)
print()
digits = ser3.str.extractall(r"(\d)")
print(digits)
print()
digits = ser3.str.extractall(r"(\d)").groupby(level=0).agg("".join)
print(digits)
```
