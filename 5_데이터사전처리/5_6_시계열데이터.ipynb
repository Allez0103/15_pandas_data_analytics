```python
import pandas as pd

dates = ['2019-01-01', '2020-03-01', '2026-06-01']

ts_dates = pd.to_datetime(dates)
print(ts_dates)
```

```python
pr_day = ts_dates.to_period(freq='D')
pr_month = ts_dates.to_period(freq='M')
pr_year = ts_dates.to_period(freq='Y')
print(f"{pr_day}\n{pr_month}\n{pr_year}")
```

```python
# 시계열 데이터 만들기

import pandas as pd

ts_ms = pd.date_range(
    start='2024-01-01',
    end=None,
    periods=6,
    freq='MS',
    tz='Asia/Seoul'
)
print(ts_ms)
```

```python
ts_me = pd.date_range(
    start='2024-01-01',
    end=None,
    periods=6,
    freq='ME',
    tz='Asia/Seoul'
)
print(ts_me)
```

```python
ts_3ms = pd.date_range(
    start='2024-01-01',
    end=None,
    periods=6,
    freq='3MS',
    tz='Asia/Seoul'
)
print(ts_3ms)
```

```python
pr_m = pd.period_range(
    start='2024-01-01',
    end=None,
    periods=3,
    freq="M"
)
print(pr_m)
```

```python
pr_h = pd.period_range(
    start='2024-01-01',
    end=None,
    periods=3,
    freq="h"
)
print(pr_h)
```

```python
# 시계열 데이터 활용
# 날짜 데이터 분리

df = pd.read_csv('./data/stock-data.csv')
df['new_Date'] = pd.to_datetime(df['Date'])

df['Year'] = df['new_Date'].dt.year
df['Month'] = df['new_Date'].dt.month
df['Day'] = df['new_Date'].dt.day

df['Date_m'] = df['new_Date'].dt.to_period(freq='M')
df.head()
```
