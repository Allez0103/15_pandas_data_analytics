```python
# 단위 환산

import pandas as pd

df = pd.read_csv('./data/auto-mpg.csv', header=None)
df.columns = [
    "mpg", "cylinders", "displacement", "horsepower", "weight", "acceleration", "model year", "origin", "name"
]

df.head()
```

```python
mpg_to_kpl = 1.60934 / 3.78541
df['kpl'] = df['mpg'] * mpg_to_kpl
df.head(3)
```

```python
df['kpl'] = df['kpl'].round(2)
df.head()
```

```python
# 자료형 변환

df.dtypes
```

```python
df['horsepower'].unique()
```

```python
import numpy as np

df['horsepower'] = df['horsepower'].replace('?', np.nan)        # str.replace -> 문자열 안의 요소를 찾아내서 대체
df = df.dropna(subset=['horsepower'], axis=0)
df['horsepower'] = df['horsepower'].astype('float')
df['horsepower'].unique()
```

```python
df['horsepower'].dtype
```

```python
print(df['origin'].unique())
print(df['origin'].nunique())       # unique 값이 몇개인지 반환
```

```python
df['origin'] = df['origin'].replace(
    {
        1 : 'USA',
        2 : 'EU',
        3 : 'JAPAN'
    }
)

# print(df['origin'].unique())
df['origin'] = df['origin'].astype('category')
print(df.info())
```

```python
df['model year'] = df['model year'].astype('category')
print(df['model year'].sample(3))
print(df['model year'].dtypes)
```
